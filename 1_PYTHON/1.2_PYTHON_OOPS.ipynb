## Introduction to OOPS - Notes

### Basic Syntax and Definitions

#### 1. Class Definition and Object Creation

-   **Class**: A blueprint for creating objects. It defines the attributes (data) and methods (functions) that objects of that class will have.
-   **Object (Instance)**: A specific instance of a class. When a class is defined, no memory is allocated until an object is created from it.

```python
# Class Definition
class MyClass:
    # The __init__ method is a special method (constructor) that is called when an object is created.
    # 'self' refers to the instance of the class itself.
    def __init__(self):
        print("MyClass created")

# Object Creation (Instantiation)
obj1 = MyClass()
# Output: MyClass created
```

#### 2. Instance Method

-   **Instance Method**: A function defined inside a class that operates on an instance of the class. It always takes `self` as its first parameter.

```python
class MyClass:
    def __init__(self):
        print("MyClass created")

    # Instance method
    def greet(self):
        print("Hello from MyClass!")

# Create an object
obj1 = MyClass()
# Output: MyClass created

# Call the instance method
obj1.greet()
# Output: Hello from MyClass!
```

### Practical Application

#### 1. Car Class

```python
class Car:
    # Class attribute: Shared by all instances of the class
    number_of_wheels = 4

    def __init__(self, make, model, year):
        # Instance attributes: Unique to each instance
        self.make = make
        self.model = model
        self.year = year
        print(f"A new {self.make} {self.model} ({self.year}) has been created.")

    # Instance method to display car information
    def display_info(self):
        print(f"Car: {self.year} {self.make} {self.model}")

# Create Car objects
car1 = Car("Toyota", "Camry", 2020)
car2 = Car("Honda", "Civic", 2022)

# Call display_info for both cars
car1.display_info()
car2.display_info()

# Accessing class attribute
print(f"Car.number_of_wheels: {Car.number_of_wheels}") # Access via class
print(f"car1.number_of_wheels: {car1.number_of_wheels}") # Access via object

# Output:
# A new Toyota Camry (2020) has been created.
# A new Honda Civic (2022) has been created.
# Car: 2020 Toyota Camry
# Car: 2022 Honda Civic
# Car.number_of_wheels: 4
# car1.number_of_wheels: 4
```

#### 2. BankAccount Class

```python
class BankAccount:
    def __init__(self, account_holder):
        self.account_holder = account_holder
        self.balance = 0  # Initial balance is 0
        print(f"Account created for {self.account_holder} with initial balance {self.balance}")

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"Deposited {amount}. New balance: {self.balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive.")
        elif amount > self.balance:
            print("Error: Insufficient funds.")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")

    def get_balance(self):
        return self.balance

# Create an object of BankAccount
account = BankAccount("Alice Smith")
print(f"Current balance: {account.get_balance()}")

# Deposit money
account.deposit(1000)
print(f"Current balance: {account.get_balance()}")

# Withdraw money
account.withdraw(300)
print(f"Current balance: {account.get_balance()}")

# Attempt to withdraw more than balance (should fail)
account.withdraw(1500)
print(f"Current balance: {account.get_balance()}")

# Output:
# Account created for Alice Smith with initial balance 0
# Current balance: 0
# Deposited 1000. New balance: 1000
# Current balance: 1000
# Withdrew 300. New balance: 700
# Current balance: 700
# Error: Insufficient funds.
# Current balance: 700
```

## Encapsulation - Notes

### Basic Level

**Encapsulation** is the bundling of data (attributes) and methods that operate on the data into a single unit or class. It restricts direct access to some of an object's components, which can prevent accidental modification of data. In Python, encapsulation is primarily achieved through naming conventions and properties.

#### 1. Protected Attribute (`_attribute`)

-   Attributes prefixed with a **single underscore** (`_`) are considered "protected." This is a convention among Python developers indicating that the attribute should be treated as non-public and not directly accessed from outside the class, though it's technically still accessible.

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self._price = price  # Protected attribute

    def get_price(self):
        print(f"Accessing _price from inside method: {self._price}")
        return self._price

# Create an object
product1 = Product("Laptop", 1200)

# Accessing _price via method (recommended)
product1.get_price()

# Accessing _price directly (discouraged, but possible)
print(f"Accessing _price directly from instance: {product1._price}")

# Output:
# Accessing _price from inside method: 1200
# Accessing _price directly from instance: 1200
```

#### 2. Private Attribute (`__attribute`)

-   Attributes prefixed with a **double underscore** (`__`) are considered "private." Python implements a mechanism called **name mangling** for these attributes. The name is internally changed to `_ClassName__attribute_name` to make it harder to access directly from outside the class, but it's not truly private like in some other languages.

```python
class User:
    def __init__(self, username, password):
        self.username = username
        self.__password = password  # Private attribute

    def display_username(self):
        print(f"Username: {self.username}")

    def _get_password_internal(self):
        # Accessing private attribute from inside the class
        return self.__password

# Create an object
user1 = User("john_doe", "mysecretpwd")

user1.display_username()

# Attempt to access __password directly (will raise AttributeError)
try:
    print(user1.__password)
except AttributeError as e:
    print(f"Attempted direct access failed: {e}")

# Demonstrating name mangling: Accessing via the mangled name (discouraged)
# The name becomes _User__password
print(f"Accessing mangled name: {user1._User__password}")

# Output:
# Username: john_doe
# Attempted direct access failed: 'User' object has no attribute '__password'
# Accessing mangled name: mysecretpwd
```

### Intermediate Level

#### 1. Single Underscore (`_attribute`) Usage

-   As shown above, `_attribute` is a convention. It signals to other developers "don't touch this directly," but Python doesn't prevent access.

```python
class Configuration:
    def __init__(self):
        self._settings = {
            "theme": "dark",
            "language": "en",
            "version": "1.0"
        }

    def get_setting(self, key):
        return self._settings.get(key, "Setting not found")

# Create an object
config = Configuration()

# Safely retrieve a setting
print(f"Current theme: {config.get_setting('theme')}")

# Direct access (discouraged, but possible)
print(f"Directly accessing all settings: {config._settings}")

# Modify directly (can lead to unexpected behavior if not careful)
config._settings["theme"] = "light"
print(f"Modified theme via direct access: {config.get_setting('theme')}")

# Output:
# Current theme: dark
# Directly accessing all settings: {'theme': 'dark', 'language': 'en', 'version': '1.0'}
# Modified theme via direct access: light
```

#### 2. Double Underscore (`__attribute`) and Name Mangling in Practice

-   Name mangling prevents accidental name clashes in subclasses and provides a weak form of privacy.

```python
class DataProcessor:
    def __init__(self, data):
        self.__internal_data = list(data)  # Private attribute

    def process(self):
        # Accessing __internal_data from inside the class
        processed_data = [x * 2 for x in self.__internal_data]
        print(f"Processed data: {processed_data}")
        return processed_data

# Create an object
processor = DataProcessor([1, 2, 3])
processor.process()

# Attempting to access mangled name directly
try:
    print(f"Directly accessing mangled name: {processor._DataProcessor__internal_data}")
    processor._DataProcessor__internal_data.append(4) # Can still modify if you know the mangled name
    processor.process()
except AttributeError as e:
    print(f"Error: {e}")

# Output:
# Processed data: [2, 4, 6]
# Directly accessing mangled name: [1, 2, 3]
# Processed data: [2, 4, 6, 8]
```

#### 3. Getters and Setters - Basic Implementation

-   Getters (`get_...`) and setters (`set_...`) are methods used to access and modify attributes, respectively. They allow for validation, computed values, or controlled access.

```python
class Student:
    def __init__(self, name):
        self.name = name
        self._grade = 0  # Protected attribute for grade

    def get_grade(self):
        return self._grade

    def set_grade(self, new_grade):
        if 0 <= new_grade <= 100:
            self._grade = new_grade
            print(f"Grade for {self.name} set to {new_grade}")
        else:
            print(f"Error: Invalid grade {new_grade}. Grade must be between 0 and 100.")

# Create a Student object
student1 = Student("John Doe")
print(f"Initial grade: {student1.get_grade()}")

# Set valid grade
student1.set_grade(85)
print(f"New grade: {student1.get_grade()}")

# Attempt to set invalid grades
student1.set_grade(105)
student1.set_grade(-5)
print(f"Grade after invalid attempts: {student1.get_grade()}")

# Output:
# Initial grade: 0
# Grade for John Doe set to 85
# New grade: 85
# Error: Invalid grade 105. Grade must be between 0 and 100.
# Error: Invalid grade -5. Grade must be between 0 and 100.
# Grade after invalid attempts: 85
```

### Advanced Level

#### 1. `@property` Decorator for Encapsulation

-   The `@property` decorator provides a Pythonic way to implement getters, setters, and deleters, allowing you to access methods like attributes, improving readability and maintainability.

```python
class Student:
    def __init__(self, name):
        self.name = name
        self._grade = 0  # Internal storage for grade

    @property
    def grade(self):
        # This method acts as the getter for 'grade'
        print(f"Getting grade for {self.name}")
        return self._grade

    @grade.setter
    def grade(self, new_grade):
        # This method acts as the setter for 'grade'
        print(f"Attempting to set grade for {self.name} to {new_grade}")
        if 0 <= new_grade <= 100:
            self._grade = new_grade
            print(f"Grade updated to {new_grade}")
        else:
            print(f"Error: Invalid grade {new_grade}. Grade must be between 0 and 100.")

# Create a Student object
student1 = Student("Jane Doe")

# Access grade using the property (getter is called)
print(f"Initial grade: {student1.grade}")

# Set grade using the property (setter is called)
student1.grade = 92
print(f"Current grade: {student1.grade}")

# Attempt to set invalid grade
student1.grade = 110
print(f"Final grade: {student1.grade}")

# Output:
# Getting grade for Jane Doe
# Initial grade: 0
# Attempting to set grade for Jane Doe to 92
# Grade updated to 92
# Getting grade for Jane Doe
# Current grade: 92
# Attempting to set grade for Jane Doe to 110
# Error: Invalid grade 110. Grade must be between 0 and 100.
# Getting grade for Jane Doe
# Final grade: 92
```

#### 2. Practical Application - Encapsulated `BankAccount` Class

```python
class SecureBankAccount:
    def __init__(self, account_number, account_holder_name, initial_balance=0):
        # Private/protected attributes
        self._account_number = account_number
        self._account_holder_name = account_holder_name
        self._balance = initial_balance
        print(f"Secure account {self._account_number} created for {self._account_holder_name}.")

    # Getter for account_number (read-only property)
    @property
    def account_number(self):
        return self._account_number

    # Getter and setter for account_holder_name
    @property
    def account_holder_name(self):
        return self._account_holder_name

    @account_holder_name.setter
    def account_holder_name(self, new_name):
        if new_name and isinstance(new_name, str):
            self._account_holder_name = new_name
            print(f"Account holder name updated to {new_name}.")
        else:
            print("Invalid name provided.")

    # Getter for balance (balance only modified via deposit/withdraw methods)
    @property
    def balance(self):
        return self._balance

    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
            print(f"Deposited {amount}. New balance: {self.balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive.")
        elif amount > self._balance:
            print("Insufficient funds.")
        else:
            self._balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")

    def display_account_details(self):
        print("\n--- Account Details ---")
        print(f"Account Number: {self.account_number}")
        print(f"Account Holder: {self.account_holder_name}")
        print(f"Current Balance: {self.balance}")
        print("-----------------------")

# Demonstrate usage
account = SecureBankAccount("12345", "Bob Johnson", 500)
account.display_account_details()

# Try valid operations
account.deposit(200)
account.withdraw(100)
account.account_holder_name = "Robert Johnson"

# Try invalid operations
account.deposit(-50)
account.withdraw(1000) # Should fail
account.account_number = "99999" # Should fail (no setter for account_number)

account.display_account_details()

# Output:
# Secure account 12345 created for Bob Johnson.
#
# --- Account Details ---
# Account Number: 12345
# Account Holder: Bob Johnson
# Current Balance: 500
# -----------------------
# Deposited 200. New balance: 700
# Withdrew 100. New balance: 600
# Account holder name updated to Robert Johnson.
# Deposit amount must be positive.
# Insufficient funds.
# Traceback (most recent call last):
#   File "<stdin>", line 49, in <module>
# AttributeError: property 'account_number' of 'SecureBankAccount' object has no setter
```

## Inheritance - Notes

### Basic Level

**Inheritance** is a mechanism in object-oriented programming that allows a new class (subclass/child class) to inherit properties and behaviors (attributes and methods) from an existing class (superclass/parent class). This promotes code reusability and establishes a natural hierarchy between related classes.

#### 1. Single Inheritance - Animal Hierarchy

-   A class inherits from a single base class.

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound")

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name) # Call parent class constructor
        self.breed = breed

    # Method overriding: Dog provides its own implementation of speak()
    def speak(self):
        print(f"{self.name} barks!")

# Instantiate Animal and Dog objects
generic_animal = Animal("Generic Animal")
my_dog = Dog("Buddy", "Golden Retriever")

# Call their speak() methods
generic_animal.speak() # Output: Generic Animal makes a sound
my_dog.speak()         # Output: Buddy barks!
```

#### 2. Accessing Parent Members

-   Child classes automatically have access to public and protected attributes and methods of their parent class.

```python
class Parent:
    def __init__(self, value):
        self.value = value

    def get_value(self):
        return self.value

class Child(Parent):
    def __init__(self, value, extra_data):
        super().__init__(value) # Initialize parent's attributes
        self.extra_data = extra_data

    def display_child_info(self):
        # Accessing parent's attribute and method
        print(f"Child's value from parent: {self.value}")
        print(f"Child calling parent's get_value(): {self.get_value()}")
        print(f"Child's extra data: {self.extra_data}")

# Create a Child object
child_obj = Child(100, "additional info")
child_obj.display_child_info()

# Direct access to parent's members from child object
print(f"Directly accessing child_obj.value: {child_obj.value}")
print(f"Directly calling child_obj.get_value(): {child_obj.get_value()}")

# Output:
# Child's value from parent: 100
# Child calling parent's get_value(): 100
# Child's extra data: additional info
# Directly accessing child_obj.value: 100
# Directly calling child_obj.get_value(): 100
```

### Intermediate Level

#### 1. Multiple Inheritance

-   A class can inherit from multiple base classes, combining their functionalities.

```python
class Flyer:
    def fly(self):
        print("I can fly!")

class Swimmer:
    def swim(self):
        print("I can swim!")

class Duck(Flyer, Swimmer):
    def __init__(self, name):
        self.name = name
        print(f"Hi, I'm {self.name}, a duck!")

# Create a Duck object
lucky_duck = Duck("Donald")

# Call methods from both base classes
lucky_duck.fly()
lucky_duck.swim()

# Output:
# Hi, I'm Donald, a duck!
# I can fly!
# I can swim!
```

#### 2. Multi-Level Inheritance

-   A child class inherits from a parent class, which in turn inherits from another parent class, forming a chain.

```python
class Vehicle:
    def __init__(self, max_speed):
        self.max_speed = max_speed

    def display_info(self):
        print(f"Max Speed: {self.max_speed} km/h")

class Car(Vehicle):
    def __init__(self, max_speed, num_doors):
        super().__init__(max_speed) # Call Vehicle's __init__
        self.num_doors = num_doors

    def display_info(self):
        super().display_info() # Call Vehicle's display_info
        print(f"Number of Doors: {self.num_doors}")

class SportsCar(Car):
    def __init__(self, max_speed, num_doors, turbo_enabled):
        super().__init__(max_speed, num_doors) # Call Car's __init__
        self.turbo_enabled = turbo_enabled

    def display_info(self):
        super().display_info() # Call Car's display_info
        print(f"Turbo Enabled: {self.turbo_enabled}")

# Create a SportsCar object
my_sports_car = SportsCar(250, 2, True)
my_sports_car.display_info()

# Output:
# Max Speed: 250 km/h
# Number of Doors: 2
# Turbo Enabled: True
```

#### 3. Hierarchical Inheritance

-   Multiple subclasses inherit from a single base class.

```python
import math

class Shape:
    def get_area(self):
        return 0 # Generic implementation

class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def get_area(self):
        return math.pi * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def get_area(self):
        return self.width * self.height

# Instantiate objects
circle = Circle(5)
rectangle = Rectangle(4, 6)

# Call get_area() polymorphically
print(f"Circle Area: {circle.get_area():.2f}") # Output: Circle Area: 78.54
print(f"Rectangle Area: {rectangle.get_area()}") # Output: Rectangle Area: 24
```

#### 4. Method Overriding with `super()`

-   A subclass provides a specific implementation for a method that is already defined in its parent class. `super()` is used to call a method from the parent class.

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    def display_info(self):
        print(f"Name: {self.name}, Salary: ${self.salary}")

class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary) # Call parent's __init__
        self.department = department

    # Override display_info()
    def display_info(self):
        super().display_info() # Call parent's display_info
        print(f"Department: {self.department}")

# Instantiate objects
emp = Employee("Alice", 50000)
manager = Manager("Bob", 75000, "Sales")

emp.display_info()
print("\n")
manager.display_info()

# Output:
# Name: Alice, Salary: $50000
#
# Name: Bob, Salary: $75000
# Department: Sales
```

### Advanced Level

#### 1. `super()` with Multiple Inheritance and MRO

-   **Method Resolution Order (MRO)**: In multiple inheritance, Python determines the order in which base classes are searched for methods using the C3 linearization algorithm. `super()` respects this MRO.

```python
class A:
    def method(self):
        print("Method from A")

class B(A):
    def method(self):
        print("Method from B")
        super().method() # Calls A.method() based on MRO

class C(A):
    def method(self):
        print("Method from C")
        super().method() # Calls A.method() based on MRO

class D(B, C):
    def method(self):
        print("Method from D")
        super().method() # Calls B.method() then C.method() then A.method() based on MRO

# Create an instance of D and call its method
d_obj = D()
d_obj.method()

# Inspect the MRO
print("\nMRO of D:")
print(D.__mro__)
# Output:
# Method from D
# Method from B
# Method from C
# Method from A
#
# MRO of D:
# (<class '__main__.D'>, <class '__main__.B'>, <class '__main__.C'>, <class '__main__.A'>, <class 'object'>)
```

#### 2. Practical Application - Library Management System Hierarchy

```python
class LibraryItem:
    def __init__(self, title, publication_year):
        self.title = title
        self.publication_year = publication_year

    def display_details(self):
        print(f"Title: {self.title}, Year: {self.publication_year}")

class Book(LibraryItem):
    def __init__(self, title, publication_year, author, isbn):
        super().__init__(title, publication_year)
        self.author = author
        self.isbn = isbn

    def display_details(self):
        super().display_details()
        print(f"  Author: {self.author}, ISBN: {self.isbn}")

class Magazine(LibraryItem):
    def __init__(self, title, publication_year, issue_number):
        super().__init__(title, publication_year)
        self.issue_number = issue_number

    def display_details(self):
        super().display_details()
        print(f"  Issue No: {self.issue_number}")

class User:
    def __init__(self, name, user_id):
        self.name = name
        self.user_id = user_id

    def display_user_info(self):
        print(f"User: {self.name} (ID: {self.user_id})")

class Borrower(User):
    def __init__(self, name, user_id):
        super().__init__(name, user_id)
        self.borrowed_items = []

    def borrow_item(self, item):
        if item not in self.borrowed_items:
            self.borrowed_items.append(item)
            print(f"{self.name} borrowed '{item.title}'.")
        else:
            print(f"{self.name} already has '{item.title}'.")

    def return_item(self, item):
        if item in self.borrowed_items:
            self.borrowed_items.remove(item)
            print(f"{self.name} returned '{item.title}'.")
        else:
            print(f"{self.name} did not borrow '{item.title}'.")

    def list_borrowed_items(self):
        print(f"\nItems borrowed by {self.name}:")
        if not self.borrowed_items:
            print("  (None)")
        for item in self.borrowed_items:
            item.display_details()

# Create library items
book1 = Book("The Python Journey", 2023, "P. Dev", "978-1234567890")
magazine1 = Magazine("Tech Weekly", 2024, 52)
book2 = Book("OOP Masterclass", 2022, "A. Coder", "978-0987654321")

# Create a borrower
bob = Borrower("Bob", "B001")

# Demonstrate borrowing and returning
bob.borrow_item(book1)
bob.borrow_item(magazine1)
bob.list_borrowed_items()

bob.borrow_item(book1) # Already borrowed
bob.return_item(book2) # Not borrowed

bob.return_item(book1)
bob.list_borrowed_items()

# Output:
# Bob borrowed 'The Python Journey'.
# Bob borrowed 'Tech Weekly'.
#
# Items borrowed by Bob:
# Title: The Python Journey, Year: 2023
#   Author: P. Dev, ISBN: 978-1234567890
# Title: Tech Weekly, Year: 2024
#   Issue No: 52
# Bob already has 'The Python Journey'.
# Bob did not borrow 'OOP Masterclass'.
# Bob returned 'The Python Journey'.
#
# Items borrowed by Bob:
# Title: Tech Weekly, Year: 2024
#   Issue No: 52
```

## Polymorphism - Notes

### Basic Level

**Polymorphism** means "many forms." In OOP, it refers to the ability of different objects to respond to the same method call in their own way. This often involves method overriding, where subclasses provide their specific implementation of a method defined in a superclass, or duck typing in Python, where the type of an object is less important than what methods it defines.

#### 1. Method Overriding with Animals

-   Subclasses provide their unique implementation of a method inherited from a parent class.

```python
class Animal:
    def __init__(self, name):
        self.name = name

    def make_sound(self):
        print(f"{self.name} makes a generic sound.")

class Dog(Animal):
    def make_sound(self):
        print(f"{self.name} barks!")

class Cat(Animal):
    def make_sound(self):
        print(f"{self.name} meows!")

# Create objects of different animal types
animal = Animal("Beast")
dog = Dog("Buddy")
cat = Cat("Whiskers")

# Store them in a list
animals = [animal, dog, cat]

# Iterate and call make_sound() polymorphically
for creature in animals:
    creature.make_sound()

# Output:
# Beast makes a generic sound.
# Buddy barks!
# Whiskers meows!
```

### Intermediate Level

#### 1. Duck Typing Example

-   In Python, if an object walks like a duck and quacks like a duck, it's a duck! This means we care more about *what an object can do* (its methods) rather than *what type it is*.

```python
class Duck:
    def quack(self):
        print("Quack! Quack!")

class Person:
    def quack(self):
        print("I can quack like a duck! (Quack!)")

class Robot:
    def quack(self):
        print("[System beep] Initiating quack sequence... Quack.")

def make_it_quack(obj):
    # This function doesn't care about the object's type,
    # only that it has a 'quack' method.
    obj.quack()

# Create instances
duck_obj = Duck()
person_obj = Person()
robot_obj = Robot()

# Pass different objects to the same function
make_it_quack(duck_obj)
make_it_quack(person_obj)
make_it_quack(robot_obj)

# Output:
# Quack! Quack!
# I can quack like a duck! (Quack!)
# [System beep] Initiating quack sequence... Quack.
```

#### 2. Polymorphism in Document Processing

-   Different document types can be processed uniformly through a common interface/method.

```python
import abc

class Document(abc.ABC):
    def __init__(self, filename):
        self.filename = filename

    @abc.abstractmethod
    def read_content(self):
        raise NotImplementedError("Subclasses must implement read_content()")

class PDFDocument(Document):
    def read_content(self):
        print(f"Reading PDF content from {self.filename}")
        # Simulate reading PDF-specific content
        return f"PDF data from {self.filename}"

class WordDocument(Document):
    def read_content(self):
        print(f"Reading Word document content from {self.filename}")
        # Simulate reading Word-specific content
        return f"Word data from {self.filename}"

class TextDocument(Document):
    def read_content(self):
        print(f"Reading plain text content from {self.filename}")
        # Simulate reading text-specific content
        return f"Text data from {self.filename}"

def process_documents(documents):
    print("\n--- Processing Documents ---")
    for doc in documents:
        doc.read_content()
    print("--------------------------")

# Create a list of different document types
documents_to_process = [
    PDFDocument("report.pdf"),
    WordDocument("memo.docx"),
    TextDocument("notes.txt"),
    PDFDocument("invoice.pdf")
]

# Process all documents using the polymorphic function
process_documents(documents_to_process)

# Output:
#
# --- Processing Documents ---
# Reading PDF content from report.pdf
# Reading Word document content from memo.docx
# Reading plain text content from notes.txt
# Reading PDF content from invoice.pdf
# --------------------------
```

```python
class Weapon:
    def __init__(self, name, damage):
        self.name = name
        self.damage = damage

    def use(self, character_name):
        print(f"  {character_name} uses {self.name} for {self.damage} damage.")

class Character:
    def __init__(self, name, weapon=None):
        self.name = name
        self.weapon = weapon # Composition: Character 'has-a' Weapon

    def attack(self, target):
        print(f"{self.name} prepares to attack {target}.")
        if self.weapon:
            self.weapon.use(self.name)
        else:
            print(f"  {self.name} performs a basic attack on {target}.")

class Warrior(Character):
    def attack(self, target):
        print(f"{self.name} lunges at {target} with great force!")
        if self.weapon:
            self.weapon.use(self.name)
        else:
            print(f"  {self.name} swings their fists at {target}.")

class Mage(Character):
    def attack(self, target):
        print(f"{self.name} casts a spell upon {target}!")
        if self.weapon:
            self.weapon.use(self.name)
        else:
            print(f"  {self.name} zaps {target} with arcane energy.")

class Archer(Character):
    def attack(self, target):
        print(f"{self.name} aims carefully at {target}.")
        if self.weapon:
            self.weapon.use(self.name)
        else:
            print(f"  {self.name} fires an arrow at {target}.")

# Create weapons
sword = Weapon("Sword of Valor", 25)
staff = Weapon("Staff of Wisdom", 30)
bow = Weapon("Longbow of Precision", 20)

# Create characters with and without weapons
warrior = Warrior("Arthur", sword)
mage = Mage("Merlin", staff)
archer = Archer("Legolas", bow)
rogue = Character("Kael") # A rogue without a specific weapon

# List of characters
characters = [warrior, mage, archer, rogue]

# Demonstrate polymorphic attacks
print("--- Round 1 Attacks ---")
for char in characters:
    char.attack("Goblin")
    print("\n")

# Change a weapon and re-attack
print("--- Round 2 Attacks (Mage gets new weapon) ---")
mage.weapon = Weapon("Orb of Fire", 40)
for char in characters:
    char.attack("Orc")
    print("\n")

# Output:
# --- Round 1 Attacks ---
# Arthur lunges at Goblin with great force!
#   Arthur uses Sword of Valor for 25 damage.
#
# Merlin casts a spell upon Goblin!
#   Merlin uses Staff of Wisdom for 30 damage.
#
# Legolas aims carefully at Goblin.
#   Legolas uses Longbow of Precision for 20 damage.
#
# Kael prepares to attack Goblin.
#   Kael performs a basic attack on Goblin.
#
#
# --- Round 2 Attacks (Mage gets new weapon) ---
# Arthur lunges at Orc with great force!
#   Arthur uses Sword of Valor for 25 damage.
#
# Merlin casts a spell upon Orc!
#   Merlin uses Orb of Fire for 40 damage.
#
# Legolas aims carefully at Orc.
#   Legolas uses Longbow of Precision for 20 damage.
#
# Kael prepares to attack Orc.
#   Kael performs a basic attack on Orc.
#
#
```

## Abstraction - Notes

### Basic Level

**Abstraction** means hiding the complex implementation details and showing only the essential features of an object. In Python, the `abc` (Abstract Base Classes) module helps to define abstract classes and methods, enforcing that subclasses provide concrete implementations.

#### 1. Defining an Abstract Base Class (ABC)

-   An ABC cannot be instantiated directly. It serves as a blueprint for other classes.
-   Use `abc.ABC` as the base class and `@abc.abstractmethod` decorator for abstract methods.

```python
import abc

class AbstractShape(abc.ABC):
    @abc.abstractmethod
    def calculate_area(self):
        pass # Abstract methods typically have no implementation

# Attempting to instantiate AbstractShape directly will raise a TypeError
try:
    shape = AbstractShape()
except TypeError as e:
    print(f"Error instantiating AbstractShape: {e}")

# Output:
# Error instantiating AbstractShape: Can't instantiate abstract class AbstractShape with abstract methods calculate_area
```

#### 2. Concrete Subclass of an ABC

-   A concrete subclass must implement all abstract methods defined in its ABC to be instantiable.

```python
import abc
import math

class AbstractShape(abc.ABC):
    @abc.abstractmethod
    def calculate_area(self):
        pass

class Circle(AbstractShape):
    def __init__(self, radius):
        self.radius = radius

    # Implement the abstract method
    def calculate_area(self):
        return math.pi * self.radius ** 2

# Instantiate a concrete Circle object
circle = Circle(radius=7)
print(f"Area of Circle: {circle.calculate_area():.2f}")

# Output:
# Area of Circle: 153.94
```

### Intermediate Level

#### 1. Enforcement of Abstract Methods

-   Python's `abc` module strictly enforces that any concrete class inheriting from an ABC *must* implement all its abstract methods. If not, the subclass itself remains abstract and cannot be instantiated.

```python
import abc

class AbstractShape(abc.ABC):
    @abc.abstractmethod
    def calculate_area(self):
        pass

    @abc.abstractmethod
    def get_perimeter(self):
        pass

class Square(AbstractShape): # Inherits from AbstractShape
    def __init__(self, side):
        self.side = side

    def calculate_area(self):
        return self.side ** 2

    # !!! Forgot to implement get_perimeter() !!!

# Attempting to instantiate Square will raise a TypeError
try:
    square = Square(5)
except TypeError as e:
    print(f"Error instantiating Square: {e}")

# Output:
# Error instantiating Square: Can't instantiate abstract class Square with abstract methods get_perimeter
```

#### 2. Registering Virtual Subclasses

-   You can register a class as a "virtual subclass" of an ABC even if it doesn't explicitly inherit from it. This is useful for third-party classes or when you want to certify that a class adheres to an interface without modifying its inheritance hierarchy.

```python
import abc

class MyAbstractClass(abc.ABC):
    @abc.abstractmethod
    def perform_action(self):
        pass

class MyConcreteClass:
    def perform_action(self):
        print("MyConcreteClass performing its action.")

    def another_method(self):
        print("MyConcreteClass has another method.")

# Register MyConcreteClass as a virtual subclass of MyAbstractClass
MyAbstractClass.register(MyConcreteClass)

# Now, isinstance() will return True, even though there's no explicit inheritance
concrete_obj = MyConcreteClass()
print(f"Is concrete_obj an instance of MyAbstractClass? {isinstance(concrete_obj, MyAbstractClass)}")

# Output:
# Is concrete_obj an instance of MyAbstractClass? True
```

#### 1. Complex ABC Hierarchy for Payment Processing

-   ABCs can be part of an inheritance hierarchy themselves, creating more specialized abstract interfaces.

```python
import abc

class PaymentProcessor(abc.ABC):
    @abc.abstractmethod
    def process_payment(self, amount):
        pass

    @abc.abstractmethod
    def refund_payment(self, transaction_id):
        pass

class OnlinePaymentProcessor(PaymentProcessor, abc.ABC):
    @abc.abstractmethod
    def authenticate_transaction(self, credentials):
        pass

class CreditCardProcessor(OnlinePaymentProcessor):
    def process_payment(self, amount):
        print(f"Processing credit card payment of ${amount}.")
        return f"txn_{amount}_CC"

    def refund_payment(self, transaction_id):
        print(f"Refunding transaction {transaction_id} via credit card.")
        return True

    def authenticate_transaction(self, credentials):
        print(f"Authenticating credit card transaction with credentials: {credentials[:3]}***") # Mask credentials
        return True

# Create an instance of the concrete processor
cc_processor = CreditCardProcessor()

# Demonstrate calling methods
transaction_id = cc_processor.process_payment(150.75)
cc_processor.authenticate_transaction("user123_passXYZ")
cc_processor.refund_payment(transaction_id)

# Output:
# Processing credit card payment of $150.75.
# Authenticating credit card transaction with credentials: use***
# Refunding transaction txn_150.75_CC via credit card.
```

#### 2. `@abstractproperty`, `@abstractclassmethod`, `@abstractstaticmethod`

-   These decorators allow you to define abstract properties, class methods, and static methods that must be implemented by concrete subclasses.

```python
import abc

class AdvancedFeature(abc.ABC):
    @property
    @abc.abstractmethod
    def feature_name(self): # Abstract property (getter)
        pass

    @feature_name.setter
    @abc.abstractmethod
    def feature_name(self, value): # Abstract property (setter)
        pass

    @abc.abstractclassmethod
    def create_feature(cls, config): # Abstract class method
        pass

    @abc.abstractstaticmethod
    def validate_input(data): # Abstract static method
        pass

class ConcreteFeature(AdvancedFeature):
    def __init__(self, name):
        self._feature_name = name

    @property
    def feature_name(self):
        return self._feature_name

    @feature_name.setter
    def feature_name(self, value):
        if isinstance(value, str) and len(value) > 0:
            self._feature_name = value
        else:
            raise ValueError("Feature name must be a non-empty string.")

    @classmethod
    def create_feature(cls, config):
        print(f"Creating feature from config: {config}")
        return cls(config.get('name', 'DefaultFeature'))

    @staticmethod
    def validate_input(data):
        print(f"Validating input: {data}")
        return isinstance(data, dict) and "param" in data

# Demonstrate usage
feature_config = {"name": "DataAnalytics", "param": 123}
if ConcreteFeature.validate_input(feature_config):
    my_feature = ConcreteFeature.create_feature(feature_config)
    print(f"Feature name: {my_feature.feature_name}")
    my_feature.feature_name = "NewDataAnalytics"
    print(f"Updated feature name: {my_feature.feature_name}")

try:
    my_feature.feature_name = ""
except ValueError as e:
    print(f"Error setting feature name: {e}")

# Output:
# Validating input: {'name': 'DataAnalytics', 'param': 123}
# Creating feature from config: {'name': 'DataAnalytics', 'param': 123}
# Feature name: DataAnalytics
# Updated feature name: NewDataAnalytics
# Error setting feature name: Feature name must be a non-empty string.
```

## Special Methods (Dunder Methods) - Notes

### Basic Level

**Special methods**, also known as "dunder methods" (due to their double underscores), are predefined methods in Python that allow you to define how objects of your class behave with built-in operations, functions, and language constructs (e.g., `+`, `-`, `len()`, `print()`, `with` statement). They enable powerful features like operator overloading, custom string representations, and context management.

#### 1. `__init__` for Object Initialization

-   The `__init__` method is the constructor. It's called automatically when a new object is created from a class. It's used to initialize the instance's attributes.

```python
class Person:
    def __init__(self, name, age):
        self.name = name  # Instance attribute
        self.age = age    # Instance attribute
        print(f"Person object '{self.name}' created.")

# Instantiate two Person objects
person1 = Person("Alice", 30)
person2 = Person("Bob", 25)

# Print their name and age
print(f"Person 1: {person1.name}, {person1.age} years old.")
print(f"Person 2: {person2.name}, {person2.age} years old.")

# Output:
# Person object 'Alice' created.
# Person object 'Bob' created.
# Person 1: Alice, 30 years old.
# Person 2: Bob, 25 years old.
```

#### 2. `__str__` and `__repr__` for String Representation

-   **`__str__`**: Returns a "user-friendly" string representation of an object. It's what `print()` typically uses and is meant for end-users.
-   **`__repr__`**: Returns an "unambiguous" or "developer-friendly" string representation, ideally one that could be used to recreate the object. It's often used for debugging and is what's displayed in interactive shells.

```python
class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        # User-friendly representation
        return f"'{self.title}' by {self.author}"

    def __repr__(self):
        # Developer-friendly representation (should be unambiguous)
        return f"Book('{self.title}', '{self.author}')"

# Create a Book object
book1 = Book("The Great Python Book", "John Doe")

# Demonstrate __str__ (called by print())
print(book1)

# Demonstrate __repr__ (called by repr() or in interactive shell)
print(repr(book1))

# In an interactive shell, just typing the object name calls repr()
# book1

# Output:
# 'The Great Python Book' by John Doe
# Book('The Great Python Book', 'John Doe')
```

### Intermediate Level

#### 1. Operator Overloading (`__add__`)

-   By implementing dunder methods like `__add__`, `__sub__`, `__mul__`, etc., you can define how standard operators work with objects of your custom class.

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

    def __add__(self, other):
        # Define vector addition: (x1+x2, y1+y2)
        if isinstance(other, Vector):
            return Vector(self.x + other.x, self.y + other.y)
        else:
            raise TypeError("Can only add Vector objects together.")

# Create Vector objects
v1 = Vector(1, 2)
v2 = Vector(3, 4)
v3 = Vector(-5, 0)

# Perform vector addition using the '+' operator
sum_vector = v1 + v2
print(f"{v1} + {v2} = {sum_vector}") # Output: Vector(1, 2) + Vector(3, 4) = Vector(4, 6)

sum_three_vectors = v1 + v2 + v3
print(f"{v1} + {v2} + {v3} = {sum_three_vectors}") # Output: Vector(1, 2) + Vector(3, 4) + Vector(-5, 0) = Vector(-1, 6)

# Output:
# Vector(1, 2) + Vector(3, 4) = Vector(4, 6)
# Vector(1, 2) + Vector(3, 4) + Vector(-5, 0) = Vector(-1, 6)
```

#### 2. Container Behavior (`__len__`, `__getitem__`)

-   Implementing `__len__`, `__getitem__`, `__setitem__`, `__delitem__`, and `__contains__` allows your custom objects to behave like built-in collections (e.g., lists, dictionaries).

```python
class Song:
    def __init__(self, title, artist):
        self.title = title
        self.artist = artist

    def __str__(self):
        return f"{self.title} by {self.artist}"

    def __repr__(self):
        return f"Song('{self.title}', '{self.artist}')"

class Playlist:
    def __init__(self, name):
        self.name = name
        self._songs = []

    def add_song(self, song):
        if isinstance(song, Song):
            self._songs.append(song)
        else:
            print("Can only add Song objects to the playlist.")

    # __len__ enables len(playlist)
    def __len__(self):
        return len(self._songs)

    # __getitem__ enables playlist[index]
    def __getitem__(self, index):
        return self._songs[index]

    def __str__(self):
        songs_str = '\n  '.join(str(s) for s in self._songs) if self._songs else '  (empty)'
        return f"Playlist '{self.name}' ({len(self)} songs):\n  {songs_str}"

# Create Song objects
song1 = Song("Track 1", "Artist A")
song2 = Song("Track 2", "Artist B")
song3 = Song("Track 3", "Artist C")

# Create a Playlist object
my_playlist = Playlist("My Favorites")

# Add songs
my_playlist.add_song(song1)
my_playlist.add_song(song2)
my_playlist.add_song(song3)

# Demonstrate len() and indexing
print(my_playlist)
print(f"Number of songs in playlist: {len(my_playlist)}") # Calls __len__
print(f"First song: {my_playlist[0]}") # Calls __getitem__
print(f"Last song: {my_playlist[-1]}")

# Output:
# Playlist 'My Favorites' (3 songs):
#   Track 1 by Artist A
#   Track 2 by Artist B
#   Track 3 by Artist C
# Number of songs in playlist: 3
# First song: Track 1 by Artist A
# Last song: Track 3 by Artist C
```

#### 3. Comparison Methods (`__eq__`, `__lt__`)

-   Dunder methods like `__eq__` (==), `__ne__` (!=), `__lt__` (<), `__le__` (<=), `__gt__` (>), `__ge__` (>=) allow you to define custom comparison logic for your objects.

```python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"Point({self.x}, {self.y})"

    def __repr__(self):
        return f"Point({self.x}, {self.y})"

    # __eq__ for equality (==)
    def __eq__(self, other):
        if isinstance(other, Point):
            return self.x == other.x and self.y == other.y
        return NotImplemented # Indicate that comparison with other types is not supported

    # Calculate distance from origin (helper for __lt__)
    def _distance_from_origin(self):
        return math.sqrt(self.x**2 + self.y**2)

    # __lt__ for less than (<) based on distance from origin
    def __lt__(self, other):
        if isinstance(other, Point):
            return self._distance_from_origin() < other._distance_from_origin()
        return NotImplemented

# Create Point objects
p1 = Point(1, 1)
p2 = Point(1, 1)
p3 = Point(2, 2)
p4 = Point(0, 0)

# Demonstrate equality
print(f"{p1} == {p2}: {p1 == p2}") # Output: Point(1, 1) == Point(1, 1): True
print(f"{p1} == {p3}: {p1 == p3}") # Output: Point(1, 1) == Point(2, 2): False

# Demonstrate less than (based on distance from origin)
print(f"{p4} < {p1}: {p4 < p1}") # Output: Point(0, 0) < Point(1, 1): True (0 < sqrt(2))
print(f"{p1} < {p3}: {p1 < p3}") # Output: Point(1, 1) < Point(2, 2): True (sqrt(2) < sqrt(8))
print(f"{p3} < {p1}: {p3 < p1}") # Output: Point(2, 2) < Point(1, 1): False

# Output:
# Point(1, 1) == Point(1, 1): True
# Point(1, 1) == Point(2, 2): False
# Point(0, 0) < Point(1, 1): True
# Point(1, 1) < Point(2, 2): True
# Point(2, 2) < Point(1, 1): False
```

### Advanced Level

#### 1. Callable Objects (`__call__`)

-   By implementing the `__call__` method, instances of your class can be called like regular functions.

```python
class Multiplier:
    def __init__(self, factor):
        self.factor = factor

    def __call__(self, value):
        return value * self.factor

# Create Multiplier objects (instances are callable)
double = Multiplier(2)
triple = Multiplier(3)
half = Multiplier(0.5)

# Call the objects like functions
print(f"Double 5: {double(5)}")      # Output: Double 5: 10
print(f"Triple 10: {triple(10)}")    # Output: Triple 10: 30
print(f"Half 7: {half(7)}")         # Output: Half 7: 3.5

# Output:
# Double 5: 10
# Triple 10: 30
# Half 7: 3.5
```

#### 2. Context Managers (`__enter__`, `__exit__`)

-   Dunder methods `__enter__` and `__exit__` enable objects to be used with the `with` statement, ensuring proper resource setup and teardown, even if errors occur.

```python
class FileHandler:
    def __init__(self, filename, mode):
        self.filename = filename
        self.mode = mode
        self.file = None

    def __enter__(self):
        print(f"[__enter__] Opening file {self.filename} in '{self.mode}' mode...")
        self.file = open(self.filename, self.mode)
        return self.file # Return the resource to be assigned to 'as' variable

    def __exit__(self, exc_type, exc_val, exc_tb):
        # exc_type, exc_val, exc_tb contain exception info if one occurred
        print(f"[__exit__] Closing file {self.filename}.")
        if self.file:
            self.file.close()
        if exc_type:
            print(f"  An exception occurred: {exc_type.__name__}: {exc_val}")
            # Return False to re-raise the exception, True to suppress it
        return False

# Test with no error
with FileHandler("example.txt", "w") as f:
    f.write("Hello, world!\n")
    print("Inside 'with' block: Data written.")
print("Outside 'with' block: File operations complete.\n")

# Test with an error
with FileHandler("example_error.txt", "w") as f:
    f.write("This line is written before error.\n")
    print("Inside 'with' block: About to raise an error...")
    raise ValueError("Something went wrong!") # Simulate an error

# Output:
# [__enter__] Opening file example.txt in 'w' mode...
# Inside 'with' block: Data written.
# [__exit__] Closing file example.txt.
# Outside 'with' block: File operations complete.
#
# [__enter__] Opening file example_error.txt in 'w' mode...
# Inside 'with' block: About to raise an error...
# [__exit__] Closing file example_error.txt.
#   An exception occurred: ValueError: Something went wrong!
# Traceback (most recent call last):
#   File "<stdin>", line 30, in <module>
# ValueError: Something went wrong!
```

#### 3. Attribute Access Control (`__setattr__`)

-   `__getattr__`, `__setattr__`, and `__delattr__` are powerful dunder methods that intercept attribute access. `__setattr__` is called whenever an attribute is assigned a value.

```python
class Rectangle:
    def __init__(self, width, height):
        # Call __setattr__ internally to validate initial values
        self.width = width
        self.height = height

    def __setattr__(self, name, value):
        # Custom validation logic for 'width' and 'height'
        if name in ("width", "height"):
            if not isinstance(value, (int, float)) or value <= 0:
                raise ValueError(f"{name} must be a positive number.")
        # Call the base class's __setattr__ to actually set the attribute
        # This prevents infinite recursion: self.name = value inside __setattr__ calls __setattr__ again!
        super().__setattr__(name, value)

    def area(self):
        return self.width * self.height

# Create a valid Rectangle
r1 = Rectangle(10, 5)
print(f"Rectangle area: {r1.area()}")

# Set valid attributes
r1.width = 12
print(f"New rectangle area: {r1.area()}")

# Attempt to set invalid attributes
try:
    r1.height = -3
except ValueError as e:
    print(f"Error setting height: {e}")

try:
    r1.width = "invalid"
except ValueError as e:
    print(f"Error setting width: {e}")

# Output:
# Rectangle area: 50
# New rectangle area: 60
# Error setting height: height must be a positive number.
# Error setting width: width must be a positive number.
```

#### 4. Memory Optimization with `__slots__`

-   `__slots__` allows you to explicitly declare the instance attributes (data members) that instances of your class will have. This can save memory, especially for classes with many instances, because it prevents the creation of `__dict__` for each instance.

```python
import sys

class HeavyObjectNoSlots:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

class HeavyObjectWithSlots:
    __slots__ = ('a', 'b', 'c') # Explicitly declare attributes

    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c

# Instantiate objects
obj_no_slots = HeavyObjectNoSlots(1, 2, 3)
obj_with_slots = HeavyObjectWithSlots(1, 2, 3)

# Compare memory footprints
print(f"Size of HeavyObjectNoSlots: {sys.getsizeof(obj_no_slots)} bytes")
print(f"Size of HeavyObjectWithSlots: {sys.getsizeof(obj_with_slots)} bytes")

# Access attributes
print(f"obj_no_slots.a: {obj_no_slots.a}")
print(f"obj_with_slots.a: {obj_with_slots.a}")

# Attempt to add a new attribute to obj_with_slots (will fail)
try:
    obj_with_slots.d = 4
except AttributeError as e:
    print(f"Error adding new attribute to HeavyObjectWithSlots: {e}")

# Output (sizes may vary slightly based on Python version/system):
# Size of HeavyObjectNoSlots: 56 bytes
# Size of HeavyObjectWithSlots: 40 bytes
# obj_no_slots.a: 1
# obj_with_slots.a: 1
# Error adding new attribute to HeavyObjectWithSlots: 'HeavyObjectWithSlots' object has no attribute 'd'
```

## Class Methods and Static Methods - Notes

### Basic Level

In Python, methods inside a class can be of three types: instance methods, class methods, and static methods. They differ in how they are defined, how they are called, and what type of arguments they implicitly receive.

#### 1. Instance Method

-   **Purpose**: Operates on an instance of the class. It can access and modify instance-specific data.
-   **Definition**: Takes `self` as its first parameter, which refers to the instance itself.
-   **Calling**: Called on an instance (e.g., `obj.method()`).

```python
class MyInstanceClass:
    def __init__(self, data):
        self.data = data # Instance attribute

    def display_data(self):
        # An instance method can access instance attributes
        print(f"Instance data: {self.data}")

# Create an object
obj = MyInstanceClass("Hello Instance!")

# Call the instance method
obj.display_data()

# Output:
# Instance data: Hello Instance!
```

#### 2. Class Method

-   **Purpose**: Operates on the class itself, or on instances but without needing instance-specific data. Often used for alternative constructors or modifying class-level state.
-   **Definition**: Decorated with `@classmethod`. Takes `cls` as its first parameter, which refers to the class itself.
-   **Calling**: Can be called on the class (e.g., `Class.method()`) or an instance (e.g., `obj.method()`).

```python
class MyClassMethod:
    count = 0  # Class attribute

    def __init__(self):
        # Instance creation can increment a class-level counter
        MyClassMethod.increment_count()

    @classmethod
    def increment_count(cls):
        # A class method can access and modify class attributes
        cls.count += 1
        print(f"Count incremented. Current class count: {cls.count}")

# Call increment_count() using the class name
MyClassMethod.increment_count()

# Create instances (which also call increment_count in __init__)
obj1 = MyClassMethod()
obj2 = MyClassMethod()

# Call increment_count() using an instance (still affects the class attribute)
obj1.increment_count()

print(f"Final class count: {MyClassMethod.count}")

# Output:
# Count incremented. Current class count: 1
# Count incremented. Current class count: 2
# Count incremented. Current class count: 3
# Count incremented. Current class count: 4
# Final class count: 4
```

#### 3. Static Method

-   **Purpose**: A utility function that logically belongs to the class but doesn't operate on the instance (`self`) or the class (`cls`). It's like a regular function but enclosed within the class's namespace.
-   **Definition**: Decorated with `@staticmethod`. Takes no special first parameter (`self` or `cls`).
-   **Calling**: Can be called on the class (e.g., `Class.method()`) or an instance (e.g., `obj.method()`).

```python
class MyStaticMethod:
    @staticmethod
    def helper_function(arg1, arg2):
        # A static method doesn't access instance or class state
        print(f"Helper function called with {arg1} and {arg2}")
        return arg1 + arg2

# Call helper_function() using the class name
result_class = MyStaticMethod.helper_function(5, 3)
print(f"Result from class call: {result_class}")

# Create an instance (static methods can be called on instances too)
static_obj = MyStaticMethod()

# Call helper_function() using an instance
result_obj = static_obj.helper_function(10, 20)
print(f"Result from object call: {result_obj}")

# Output:
# Helper function called with 5 and 3
# Result from class call: 8
# Helper function called with 10 and 20
# Result from object call: 30
```

### Intermediate Level

#### 1. Class Method as an Alternative Constructor

-   Class methods are excellent for providing alternative ways to create instances of a class.

```python
class Person:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    def display_name(self):
        print(f"Name: {self.first_name} {self.last_name}")

    @classmethod
    def from_full_name(cls, full_name):
        # This class method acts as an alternative constructor
        parts = full_name.split(' ', 1)
        if len(parts) == 2:
            return cls(parts[0], parts[1])
        else:
            return cls(full_name, "") # Handle single name case

# Create a Person object using the regular constructor
p1 = Person("Alice", "Smith")
p1.display_name()

# Create a Person object using the class method alternative constructor
p2 = Person.from_full_name("Bob Johnson")
p2.display_name()

p3 = Person.from_full_name("Charlie")
p3.display_name()

# Output:
# Name: Alice Smith
# Name: Bob Johnson
# Name: Charlie
```

#### 2. Static Method for Utility Functionality

-   Static methods are ideal for utility functions that perform calculations or operations related to the class but don't need access to instance or class state.

```python
class Calculator:
    @staticmethod
    def add_numbers(x, y):
        return x + y

    @staticmethod
    def is_even(number):
        return number % 2 == 0

# Demonstrate calling static methods directly from the class
sum_result = Calculator.add_numbers(10, 5)
print(f"Sum: {sum_result}")

parity_check = Calculator.is_even(4)
print(f"Is 4 even? {parity_check}")

parity_check_odd = Calculator.is_even(7)
print(f"Is 7 even? {parity_check_odd}")

# Output:
# Sum: 15
# Is 4 even? True
# Is 7 even? False
```

#### 3. Interaction with Attributes

```python
class Tracker:
    total_trackers = 0  # Class attribute

    def __init__(self, name):
        self.name = name  # Instance attribute
        Tracker.total_trackers += 1

    def get_name(self):
        # Instance method: Accesses instance attribute (self.name) and class attribute (Tracker.total_trackers)
        return f"Tracker name: {self.name}, Total created: {Tracker.total_trackers}"

    @classmethod
    def get_total_trackers(cls):
        # Class method: Accesses class attribute (cls.total_trackers)
        # Cannot directly access instance attributes like self.name
        return f"Total trackers created (from class method): {cls.total_trackers}"

    @staticmethod
    def print_message():
        # Static method: Cannot directly access instance or class attributes
        print("This is a general message from the Tracker class.")
        # If it needs class attribute, it must access via class name: Tracker.total_trackers

# Create Tracker objects
t1 = Tracker("Alpha")
t2 = Tracker("Beta")

# Demonstrate instance method access
print(t1.get_name())
print(t2.get_name())

# Demonstrate class method access
print(Tracker.get_total_trackers())
print(t1.get_total_trackers()) # Can also be called on an instance

# Demonstrate static method access
Tracker.print_message()
t1.print_message()

# Output:
# Tracker name: Alpha, Total created: 1
# Tracker name: Beta, Total created: 2
# Total trackers created (from class method): 2
# Total trackers created (from class method): 2
# This is a general message from the Tracker class.
# This is a general message from the Tracker class.
```

#### 4. Calling Methods Internally

-   Instance methods can call class and static methods using `self.classmethod()` or `self.staticmethod()` (or `ClassName.classmethod()` / `ClassName.staticmethod()`).

```python
class ReportGenerator:
    REPORT_TEMPLATES = {"summary": "Summary Template", "detail": "Detail Template"}

    def __init__(self, report_type, raw_data):
        self.report_type = report_type
        self.raw_data = raw_data

    @classmethod
    def get_report_template(cls, template_name):
        return cls.REPORT_TEMPLATES.get(template_name, "Default Template")

    @staticmethod
    def format_data(data):
        # Simulate data formatting
        return [d.upper() for d in data] if isinstance(data, list) else str(data).upper()

    def generate_report(self):
        print(f"\n--- Generating {self.report_type.capitalize()} Report ---")

        # Call a class method
        template = self.get_report_template(self.report_type)
        print(f"Using template: {template}")

        # Call a static method
        formatted_data = self.format_data(self.raw_data)
        print(f"Formatted Data: {formatted_data}")

        print("Report content goes here...")
        print("------------------------------------")

# Create a report generator object
report_gen = ReportGenerator("summary", ["item1", "item2", "item3"])
report_gen.generate_report()

detail_report_gen = ReportGenerator("detail", "single_entry_data")
detail_report_gen.generate_report()

# Output:
#
# --- Generating Summary Report ---
# Using template: Summary Template
# Formatted Data: ['ITEM1', 'ITEM2', 'ITEM3']
# Report content goes here...
# ------------------------------------
#
# --- Generating Detail Report ---
# Using template: Detail Template
# Formatted Data: SINGLE_ENTRY_DATA
# Report content goes here...
# ------------------------------------
```

### Advanced Level

#### 1. Polymorphism with Class Methods and Inheritance

-   Class methods can exhibit polymorphic behavior across an inheritance hierarchy, as `cls` will refer to the actual class (or subclass) on which the method was called.

```python
class Vehicle:
    vehicle_type = "Generic Vehicle"

    @classmethod
    def create_vehicle(cls, name):
        # 'cls' here will be the actual class (Vehicle, Car, or Truck)
        print(f"Creating a {cls.vehicle_type} named '{name}'.")
        return cls(name) # Return an instance of the specific class (cls)

    def __init__(self, name):
        self.name = name

    def drive(self):
        print(f"{self.name} is driving.")

class Car(Vehicle):
    vehicle_type = "Car" # Override class attribute

    def drive(self):
        print(f"{self.name} is driving on the road.")

class Truck(Vehicle):
    vehicle_type = "Truck" # Override class attribute

    def drive(self):
        print(f"{self.name} is hauling cargo.")

# Call the class method on different classes
my_car = Car.create_vehicle("My Sedan")
my_truck = Truck.create_vehicle("Big Rig")

# The returned objects are instances of Car and Truck, respectively
my_car.drive()
my_truck.drive()

# Output:
# Creating a Car named 'My Sedan'.
# Creating a Truck named 'Big Rig'.
# My Sedan is driving on the road.
# Big Rig is hauling cargo.
```

#### 2. Factory Method Pattern with Class Methods

-   The factory method pattern uses a method (often a class method) to create objects, allowing subclasses to decide which class to instantiate. This centralizes object creation logic.

```python
class Circle:
    def __init__(self, radius):
        self.radius = radius
    def draw(self):
        print(f"Drawing Circle with radius {self.radius}")

class Square:
    def __init__(self, side):
        self.side = side
    def draw(self):
        print(f"Drawing Square with side {self.side}")

class Triangle:
    def __init__(self, base, height):
        self.base = base
        self.height = height
    def draw(self):
        print(f"Drawing Triangle with base {self.base} and height {self.height}")

class ShapeFactory:
    @classmethod
    def create_shape(cls, shape_type, *args, **kwargs):
        if shape_type == "circle":
            return Circle(*args, **kwargs)
        elif shape_type == "square":
            return Square(*args, **kwargs)
        elif shape_type == "triangle":
            return Triangle(*args, **kwargs)
        else:
            raise ValueError(f"Unknown shape type: {shape_type}")

# Demonstrate creating different shapes using the factory method
circle = ShapeFactory.create_shape("circle", radius=10)
square = ShapeFactory.create_shape("square", side=5)
triangle = ShapeFactory.create_shape("triangle", base=4, height=8)

circle.draw()
square.draw()
triangle.draw()

# Output:
# Drawing Circle with radius 10
# Drawing Square with side 5
# Drawing Triangle with base 4 and height 8
```

## Error Handling and Custom Exceptions in OOPS - Notes

### Basic Level

**Error Handling** is a crucial part of building robust software. In Python, this is primarily done using `try`, `except`, `else`, and `finally` blocks. In an OOPS context, these blocks can be effectively used within methods to manage unexpected situations gracefully.

#### 1. `try-except` in a Class Method

-   The `try` block contains code that might raise an exception.
-   The `except` block catches specific exceptions and handles them.

```python
class Calculator:
    def divide(self, numerator, denominator):
        try:
            result = numerator / denominator
            print(f"Division result: {result}")
            return result
        except ZeroDivisionError:
            print("Error: Cannot divide by zero!")
            return None
        except TypeError:
            print("Error: Both numerator and denominator must be numbers.")
            return None

# Create a Calculator object
calc = Calculator()

# Demonstrate valid division
calc.divide(10, 2)

# Demonstrate division by zero
calc.divide(10, 0)

# Demonstrate division with non-numeric input
calc.divide(10, "2")

# Output:
# Division result: 5.0
# Error: Cannot divide by zero!
# Error: Both numerator and denominator must be numbers.
```

#### 2. `finally` Block for Resource Cleanup

-   The `finally` block *always* executes, regardless of whether an exception occurred or not. This makes it ideal for cleanup operations like closing files or releasing resources.

```python
class ResourceHandler:
    def __init__(self, resource_name):
        self.resource_name = resource_name
        self._is_open = False

    def _open_resource(self):
        print(f"Resource '{self.resource_name}' opened.")
        self._is_open = True

    def _close_resource(self):
        if self._is_open:
            print(f"Resource '{self.resource_name}' closed.")
            self._is_open = False

    def process_resource(self, data, raise_error=False):
        self._open_resource()
        try:
            print(f"Processing data: {data}")
            if raise_error:
                raise ValueError(f"Simulated error during processing of '{self.resource_name}'.")
            processed_data = data.upper() # Example operation
            print(f"Processed data: {processed_data}")
            return processed_data
        except ValueError as e:
            print(f"Caught error in processing: {e}")
            return None
        finally:
            self._close_resource()

# Create a resource handler
handler = ResourceHandler("MyImportantData")

# Process without error
handler.process_resource("some_data")
print("\n")

# Process with simulated error
handler.process_resource("another_data", raise_error=True)

# Output:
# Resource 'MyImportantData' opened.
# Processing data: some_data
# Processed data: SOME_DATA
# Resource 'MyImportantData' closed.
#
# Resource 'MyImportantData' opened.
# Processing data: another_data
# Caught error in processing: Simulated error during processing of 'MyImportantData'.
# Resource 'MyImportantData' closed.
```

### Intermediate Level

#### 1. Handling Specific Exceptions in Class Methods (BankAccount)

-   **Custom Exceptions**: You can define your own exception classes by inheriting from built-in `Exception` or its subclasses. This makes error handling more semantic.

```python
class InsufficientFundsError(Exception):
    """Custom exception for insufficient funds in a bank account."""
    def __init__(self, message="Insufficient funds for withdrawal.", balance=0, amount=0):
        super().__init__(message)
        self.balance = balance
        self.amount = amount

class BankAccount:
    def __init__(self, account_holder, initial_balance=0):
        if initial_balance < 0:
            raise ValueError("Initial balance cannot be negative.")
        self.account_holder = account_holder
        self._balance = initial_balance
        print(f"Account for {self.account_holder} created with balance: {self._balance}")

    def get_balance(self):
        return self._balance

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit amount must be positive.")
        self._balance += amount
        print(f"Deposited {amount}. New balance: {self._balance}")

    def withdraw(self, amount):
        if amount <= 0:
            raise ValueError("Withdrawal amount must be positive.")
        if amount > self._balance:
            raise InsufficientFundsError(
                f"Attempted to withdraw {amount}, but only {self._balance} available.",
                balance=self._balance, amount=amount
            )
        self._balance -= amount
        print(f"Withdrew {amount}. New balance: {self.balance}")

# Create a BankAccount object
account = BankAccount("Charlie", 500)

# Demonstrate handling specific exceptions
try:
    account.deposit(200)
    account.withdraw(100)
    account.withdraw(-50) # This should raise ValueError
except ValueError as e:
    print(f"Caught a ValueError: {e}")

print(f"Balance after first operations: {account.get_balance()}\n")

try:
    account.withdraw(1000) # This should raise InsufficientFundsError
except InsufficientFundsError as e:
    print(f"Caught InsufficientFundsError: {e}")
    print(f"Details: Available {e.balance}, Attempted {e.amount}")
except ValueError as e:
    print(f"Caught a ValueError: {e}")

print(f"Final balance: {account.get_balance()}")

# Output:
# Account for Charlie created with balance: 500
# Deposited 200. New balance: 700
# Withdrew 100. New balance: 600
# Caught a ValueError: Withdrawal amount must be positive.
# Balance after first operations: 600
#
# Caught InsufficientFundsError: Attempted to withdraw 1000, but only 600 available.
# Details: Available 600, Attempted 1000
# Final balance: 600
```

#### 2. Basic Custom Exception

-   Define a simple custom exception by inheriting from `Exception` (or a more specific built-in exception like `ValueError`).

```python
class InvalidAgeError(ValueError):
    """Custom exception for invalid age values."""
    def __init__(self, age, message="Age must be between 0 and 150."):
        super().__init__(message)
        self.age = age

class Person:
    def __init__(self, name, age):
        if not isinstance(age, int) or not (0 <= age <= 150):
            raise InvalidAgeError(age)
        self.name = name
        self.age = age
        print(f"Person '{self.name}' created with age {self.age}.")

# Demonstrate creating Person objects with valid and invalid ages
try:
    p1 = Person("David", 45)
    p2 = Person("Eve", 160) # Invalid age
except InvalidAgeError as e:
    print(f"Caught InvalidAgeError for age {e.age}: {e}")

try:
    p3 = Person("Frank", -10) # Invalid age
except InvalidAgeError as e:
    print(f"Caught InvalidAgeError for age {e.age}: {e}")

# Output:
# Person 'David' created with age 45.
# Caught InvalidAgeError for age 160: Age must be between 0 and 150.
# Caught InvalidAgeError for age -10: Age must be between 0 and 150.
```

### Advanced Level

#### 1. Custom Exception Hierarchy

-   You can build a hierarchy of custom exceptions, allowing for more granular catching of related errors.

```python
import abc

class ShapeError(Exception):
    """Base exception for all shape-related errors."""
    pass

class InvalidDimensionError(ShapeError):
    """Exception raised for non-positive or invalid shape dimensions."""
    def __init__(self, dimension_name, value, message="Invalid dimension value."):
        super().__init__(f"{message} {dimension_name}: {value}")
        self.dimension_name = dimension_name
        self.value = value

class Shape(abc.ABC):
    @abc.abstractmethod
    def get_area(self):
        pass

class Rectangle(Shape):
    def __init__(self, width, height):
        if not isinstance(width, (int, float)) or width <= 0:
            raise InvalidDimensionError("width", width, "Width must be a positive number.")
        if not isinstance(height, (int, float)) or height <= 0:
            raise InvalidDimensionError("height", height, "Height must be a positive number.")
        self.width = width
        self.height = height
        print(f"Rectangle created: Width={self.width}, Height={self.height}")

    def get_area(self):
        return self.width * self.height

# Demonstrate creating valid and invalid Rectangles
try:
    rect1 = Rectangle(10, 5)
    print(f"Area of rect1: {rect1.get_area()}")

    rect2 = Rectangle(-2, 5) # Invalid width
except InvalidDimensionError as e:
    print(f"Caught InvalidDimensionError: {e}")
    print(f"Problematic dimension: {e.dimension_name}, Value: {e.value}")
except ShapeError as e: # Catching the base exception
    print(f"Caught a generic ShapeError: {e}")

print("\n")

try:
    rect3 = Rectangle(5, 0) # Invalid height
except InvalidDimensionError as e:
    print(f"Caught InvalidDimensionError: {e}")
except ShapeError as e: # Catching the base exception
    print(f"Caught a generic ShapeError: {e}")

# Output:
# Rectangle created: Width=10, Height=5
# Area of rect1: 50
# Caught InvalidDimensionError: Width must be a positive number. width: -2
# Problematic dimension: width, Value: -2
#
# Caught InvalidDimensionError: Height must be a positive number. height: 0
```
