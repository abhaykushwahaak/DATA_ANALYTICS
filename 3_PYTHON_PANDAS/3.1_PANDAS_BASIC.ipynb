# PANDAS 


Pandas is a powerful data manipulation library in Python, widely used for data analysis and data cleaning. It provides two primary data structures: Series and DataFrame. A Series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).


```python
import pandas as pd
```

## SERIES

```python
## Series
## A Pandas Series is a one-dimensional array-like object that can hold any data type. It is similar to a column in a table.
# Gives default index starting from zero 

data =[1,2,3,4]
series=pd.Series(data)
print("series \n",series)
```

```python
# Create series form dictionary
# take keys of dictionary as Index 
 
data={'a':1,'b':2,'c':3}
pd.Series(data)
```

```python
# Giving the index exclusively
data=[1,2,3,4]
index=['a','b','c','d']
pd.Series(data,index=index)
```

## DATAFRAME

```python
# DATAFRAME 

# Create a dataframe from dictionary 
data={
    'name':['abhay','sai','ranjan','reeshi'],
    'age':[22,21,23,50],
    'city':['Kanpur','Bidar','Purnia','Samastipur']
}

df=pd.DataFrame(data)
print(df)
print(type)
```

```python
# we can also change this data to arrays as well
import numpy  as np

np.array(df)
```

## DATA AGGREGATING AND GROUPING 

```python
df.head()
```

```python
# CALCULATINNG PRODUCT WISE VLAUES MEAN
# df.groupby
grouped_mean=df.groupby('Product')['Value'].mean()
print(grouped_mean) 
```

```python
# group by multiple criteria
grouped_mean=df.groupby(['Region','Product'])['Value'].sum()
print(grouped_mean) 
```

#### AGGREGATE MULTIPLE FUNCTION 

```python
# aggregate multiple fucntions
grouped_agg=df.groupby('Region')['Value'].agg(['mean','sum','count'])
print(grouped_agg)
```

##  MERGING AND JOINING DATAFRAME 

```python
### Merging and joining Dataframes
# Create sample DataFrames
df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1, 2, 3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value2': [4, 5, 6]})
```

```python
df1
```

```python
df2
```

```python
# Merge Dataframe on the 'Key Columns'
# inner join
pd.merge(df1,df2,on="Key",how="inner")
```

```python
# Outer join 
pd.merge(df1,df2,on="Key",how="outer")
```

```python
# left join
pd.merge(df1,df2,on="Key",how="left")
```

```python
# right join
pd.merge(df1,df2,on="Key",how="right")
```
